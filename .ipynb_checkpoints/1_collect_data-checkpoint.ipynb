```python
from Bio import Entrez
from datetime import datetime, timedelta
import pandas as pd

def fetch_pmids_by_date_and_keywords(min_date, max_date, keywords, email="your_email@example.com", batch_size=5000):
    """
    Retrieve PMIDs from PubMed within a date range where all keywords appear in the abstract.
    Recursively splits the date range if a query exceeds PubMed's 9,999 record limit.
    """
    Entrez.email = email

    # Build query
    keyword_query = " AND ".join([f'"{kw}"[Abstract]' for kw in keywords])
    date_query = f'("{min_date}"[Date - Publication] : "{max_date}"[Date - Publication])'
    query = f"{keyword_query} AND {date_query}"

    # First search just to get count
    handle = Entrez.esearch(db="pubmed", term=query, usehistory="y", retmax=0)
    record = Entrez.read(handle)
    handle.close()
    count = int(record["Count"])

    if count == 0:
        return []

    if count > 9999:
        # Split date range in two halves
        start_dt = datetime.strptime(min_date, "%Y/%m/%d")
        end_dt = datetime.strptime(max_date, "%Y/%m/%d")
        mid_year = (start_dt.year + end_dt.year) // 2
        mid_dt = datetime(mid_year, 12, 31)

        left_pmids = fetch_pmids_by_date_and_keywords(
            min_date=min_date,
            max_date=mid_dt.strftime("%Y/%m/%d"),
            keywords=keywords,
            email=email,
            batch_size=batch_size
        )
        right_pmids = fetch_pmids_by_date_and_keywords(
            min_date=(mid_dt + timedelta(days=1)).strftime("%Y/%m/%d"),
            max_date=max_date,
            keywords=keywords,
            email=email,
            batch_size=batch_size
        )
        return left_pmids + right_pmids

    # Otherwise safe to fetch with batching
    pmids = []
    for start in range(0, count, batch_size):
        handle = Entrez.esearch(
            db="pubmed",
            term=query,
            usehistory="y",
            retstart=start,
            retmax=batch_size
        )
        record = Entrez.read(handle)
        handle.close()
        pmids.extend(record["IdList"])

    return pmids


import os
from datetime import datetime
import pandas as pd

def merge_with_existing(new_df, parquet_path="pmids_results.parquet"):
    """
    Merge new results into an existing parquet file (if present).
    Avoids duplicates by pathogen + antibiotic + interval.
    """
    if os.path.exists(parquet_path):
        old_df = pd.read_parquet(parquet_path)
        merged_df = pd.concat([old_df, new_df], ignore_index=True)
        merged_df = merged_df.drop_duplicates(
            subset=["pathogen", "antibiotic", "start_date", "end_date"],
            keep="last"
        )
    else:
        merged_df = new_df

    merged_df.to_parquet(parquet_path, index=False)
    return merged_df


def fetch_pmids_intervals(pathogens, antibiotics, min_date, max_date, year_gap=5,
                          email="your_email@example.com", parquet_path="pmids_results.parquet",
                          save_every=50):
    """
    For each (pathogen, antibiotic), query PubMed in year_gap intervals.
    If an interval exceeds PubMed's 9,999 cap, fetch_pmids_by_date_and_keywords
    automatically subdivides further.
    Skips queries already present in parquet_path (if it exists).
    Saves to parquet every `save_every` new tasks.
    """
    # Load previous results if available
    if os.path.exists(parquet_path):
        existing_df = pd.read_parquet(parquet_path)
    else:
        existing_df = pd.DataFrame(columns=["pathogen", "antibiotic", "start_date", "end_date", "pmids", "count"])

    start_year = datetime.strptime(min_date, "%Y/%m/%d").year
    end_year = datetime.strptime(max_date, "%Y/%m/%d").year

    total_tasks = len(pathogens) * len(antibiotics)
    task_counter = 0
    new_results = []

    for ai, antibiotic in enumerate(antibiotics, start=1):
        for pi, pathogen in enumerate(pathogens, start=1):
            task_counter += 1
            task_progress = (task_counter / total_tasks) * 100

            for year in range(start_year, end_year + 1, year_gap):
                interval_start = f"{year}/01/01"
                interval_end_year = min(year + year_gap - 1, end_year)
                interval_end = f"{interval_end_year}/12/31"

                # Check if already in existing_df
                mask = (
                    (existing_df["pathogen"] == pathogen) &
                    (existing_df["antibiotic"] == antibiotic) &
                    (existing_df["start_date"] == interval_start) &
                    (existing_df["end_date"] == interval_end)
                )
                if mask.any():
                    count = int(existing_df.loc[mask, "count"].iloc[0])
                    print(
                        f"[{task_counter:>3}/{total_tasks}] "
                        f"({task_progress:5.1f}%) | "
                        f"{pathogen:<40} | {antibiotic:<20} | "
                        f"{interval_start} – {interval_end} | "
                        f"{count:5} PMIDs (cached)"
                    )
                    continue

                # Otherwise, fetch fresh
                pmids = fetch_pmids_by_date_and_keywords(
                    min_date=interval_start,
                    max_date=interval_end,
                    keywords=[antibiotic, pathogen],
                    email=email
                )

                new_results.append({
                    "pathogen": pathogen,
                    "antibiotic": antibiotic,
                    "start_date": interval_start,
                    "end_date": interval_end,
                    "pmids": pmids,
                    "count": len(pmids)
                })

                print(
                    f"[{task_counter:>3}/{total_tasks}] "
                    f"({task_progress:5.1f}%) | "
                    f"{pathogen:<40} | {antibiotic:<20} | "
                    f"{interval_start} – {interval_end} | "
                    f"{len(pmids):5} PMIDs (new)"
                )

                # Periodic save every N new results
                if len(new_results) >= save_every:
                    df_new = pd.DataFrame(new_results)
                    existing_df = merge_with_existing(df_new, parquet_path=parquet_path)
                    new_results = []  # reset buffer

    # Final save for leftovers
    if new_results:
        df_new = pd.DataFrame(new_results)
        existing_df = merge_with_existing(df_new, parquet_path=parquet_path)

    return existing_df
```
