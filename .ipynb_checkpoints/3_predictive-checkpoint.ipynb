```python
import pandas as pd

def build_taxonomy_onehot(df, taxonomy_cols=None):
    """
    Build one-hot embedding vectors for taxonomic information.
    """
    if taxonomy_cols is None:
        taxonomy_cols = ["domain", "kingdom", "phylum", "class", "order", "family"]
    
    # Generate one-hot encoding
    df_onehot = pd.get_dummies(df[taxonomy_cols], prefix=taxonomy_cols)
    
    # Store each row as a list of ints
    embeddings = df_onehot.values.astype(int).tolist()
    
    # Attach back to dataframe
    df = df.copy()
    df["taxonomy_embeddings"] = embeddings
    
    return df, df_onehot
```

```python
import numpy as np
import pandas as pd

def build_data_and_target(df):
    """
    Build feature (X) and target (Y) matrices for pathogens.
    
    - Features: log1p(counts) for all antibiotics and all start_date bins except the last.
    - Targets: log1p(counts) for the last start_date bin for each antibiotic.
    
    Args:
        df (pd.DataFrame): Must contain ["pathogen","antibiotic","start_date","count"].
    
    Returns:
        X (np.ndarray): Feature matrix (n_pathogens x (n_antibiotics * (n_bins-1))).
        Y (np.ndarray): Target matrix (n_pathogens x n_antibiotics).
        feature_df (pd.DataFrame): Wide-format features with column labels.
        target_df (pd.DataFrame): Wide-format targets with column labels.
    """
    df = df.copy()
    df["start_date"] = pd.to_datetime(df["start_date"])

    # Identify bins
    all_bins = sorted(df["start_date"].unique())
    last_bin = all_bins[-1]
    feature_bins = all_bins[:-1]

    # Pivot features
    feature_df = df[df["start_date"].isin(feature_bins)].pivot_table(
        index="pathogen",
        columns=["antibiotic", "start_date"],
        values="count",
        aggfunc="sum",
        fill_value=0
    )
    feature_df = np.log1p(feature_df)
    feature_df.columns = [f"{antibiotic}_{start.date()}" for antibiotic, start in feature_df.columns]

    # Pivot targets (last bin only)
    target_df = df[df["start_date"] == last_bin].pivot_table(
        index="pathogen",
        columns="antibiotic",
        values="count",
        aggfunc="sum",
        fill_value=0
    )
    target_df = np.log1p(target_df)
    target_df.columns = [f"{antibiotic}_target" for antibiotic in target_df.columns]

    # Convert to numpy
    X = feature_df.to_numpy()
    Y = target_df.to_numpy()

    return X, Y, feature_df, target_df


def build_amr_with_embeddings(amr_df, data_df, embedding_cols):
    """
    Build AMR features and targets, then concatenate pathogen embeddings.

    Parameters
    ----------
    amr_df : pd.DataFrame
        Must contain ["pathogen","antibiotic","start_date","count"].
    data_df : pd.DataFrame
        DataFrame with pathogen info and embedding columns.
        Must have a 'CanonicalSpecies' column matching amr_df['pathogen'].
    embedding_cols : list of str
        Embedding columns in data_df to concatenate (each row is list/array).

    Returns
    -------
    X : np.ndarray
        Concatenated features (AMR + embeddings).
    Y : np.ndarray
        Target matrix.
    feature_df : pd.DataFrame
        AMR features (wide format).
    target_df : pd.DataFrame
        AMR targets (wide format).
    embeddings_aligned : np.ndarray
        Pure embeddings aligned to pathogens in amr_df.
    """
    # Step 1: build AMR features/targets
    X_amr, Y, feature_df, target_df = build_data_and_target(amr_df)

    # Step 2: collect embeddings
    pathogens = feature_df.index
    data_sub = data_df.set_index("CanonicalSpecies").loc[pathogens]

    embeddings_all = []
    for col in embedding_cols:
        emb = np.vstack(data_sub[col].apply(np.array).values)
        embeddings_all.append(emb)

    embeddings_aligned = np.hstack(embeddings_all)

    # Step 3: concatenate
    X = np.hstack([X_amr, embeddings_aligned])

    return X, Y, feature_df, target_df, embeddings_aligned
```

```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.multioutput import MultiOutputRegressor
from sklearn.model_selection import train_test_split
import sklearn.metrics as skmetrics

def random_forest_experiment_with_embeddings(
    amr_df,
    data_df,
    embedding_cols=None,
    n_repeats=5,
    test_size=0.2,
    random_state=42,
    n_estimators=200,
    max_depth=None,
):
    """
    Run repeated train/test experiments with RandomForest on pathogen AMR features
    optionally concatenated with pathogen embeddings.

    Args:
        amr_df (pd.DataFrame): Must contain ["pathogen","antibiotic","start_date","count"].
        data_df (pd.DataFrame): Must contain pathogen info + embedding columns.
        embedding_cols (list of str or None): Embedding columns to use. If None, only AMR features are used.
        n_repeats (int): Number of random splits.
        test_size (float): Fraction for test.
        random_state (int): Seed base.
        n_estimators (int): Number of trees in RF.
        max_depth (int or None): Max depth of trees.

    Returns:
        results (pd.DataFrame): Performance per repeat (R², RMSE).
        avg_scores (dict): Average R² and RMSE across repeats.
    """
    if embedding_cols:
        X, Y, feature_df, target_df, embeddings = build_amr_with_embeddings(amr_df, data_df, embedding_cols)
    else:
        X, Y, feature_df, target_df = build_data_and_target(amr_df)
        embeddings = None

    pathogens = feature_df.index

    # Compute weights = log1p(total count for pathogen)
    pathogen_totals = amr_df.groupby("pathogen")["count"].sum()
    pathogen_weights = np.log1p(pathogen_totals.loc[pathogens].values)

    r2_scores = []
    rmse_scores = []

    for i in range(n_repeats):
        # Train/test split
        train_idx, test_idx = train_test_split(
            np.arange(len(pathogens)), test_size=test_size, random_state=random_state + i
        )

        X_train, Y_train = X[train_idx], Y[train_idx]
        X_test, Y_test = X[test_idx], Y[test_idx]
        sample_weights = pathogen_weights[train_idx]

        # Model
        base_model = RandomForestRegressor(
            n_estimators=n_estimators,
            max_depth=max_depth,
            random_state=random_state + i,
            n_jobs=-1
        )
        model = MultiOutputRegressor(base_model)

        # Fit with weights
        model.fit(X_train, Y_train, sample_weight=sample_weights)

        # Predict
        Y_pred = model.predict(X_test)

        # Performance measures
        r2 = skmetrics.r2_score(Y_test, Y_pred, multioutput="uniform_average")
        rmse = np.sqrt(skmetrics.mean_squared_error(Y_test, Y_pred))

        r2_scores.append(r2)
        rmse_scores.append(rmse)

        print(f"Repeat {i+1}/{n_repeats}: R²={r2:.3f}, RMSE={rmse:.3f}")

    results = pd.DataFrame({"R2": r2_scores, "RMSE": rmse_scores})
    avg_scores = {
        "R2_mean": np.mean(r2_scores),
        "R2_std": np.std(r2_scores),
        "RMSE_mean": np.mean(rmse_scores),
        "RMSE_std": np.std(rmse_scores),
    }

    print("\nAverage Performance:")
    print(f"  R²   = {avg_scores['R2_mean']:.3f} ± {avg_scores['R2_std']:.3f}")
    print(f"  RMSE = {avg_scores['RMSE_mean']:.3f} ± {avg_scores['RMSE_std']:.3f}")

    return results, avg_scores
```

```python
import numpy as np
import pandas as pd
import sklearn.metrics as skmetrics

def baseline_persistence(feature_df, target_df):
    """
    Baseline: Predict next bin = last observed bin (per pathogen & antibiotic).
    
    Args:
        feature_df (pd.DataFrame): Wide-format features from build_data_and_target.
        target_df (pd.DataFrame): Wide-format targets from build_data_and_target.
    
    Returns:
        r2 (float), rmse (float)
    """
    # For each antibiotic, last observed bin is the last column (per antibiotic)
    preds = []
    for ab in target_df.columns:
        ab_name = ab.replace("_target", "")
        last_cols = [c for c in feature_df.columns if c.startswith(ab_name + "_")]
        last_col = sorted(last_cols)[-1]  # last time bin
        preds.append(feature_df[last_col].values)
    Y_pred = np.vstack(preds).T
    Y_true = target_df.to_numpy()

    r2 = skmetrics.r2_score(Y_true, Y_pred, multioutput="uniform_average")
    rmse = np.sqrt(skmetrics.mean_squared_error(Y_true, Y_pred))
    return r2, rmse


def baseline_global_average(target_df):
    """
    Baseline: Predict global average per antibiotic for all pathogens.
    
    Args:
        target_df (pd.DataFrame): Wide-format targets from build_data_and_target.
    
    Returns:
        r2 (float), rmse (float)
    """
    Y_true = target_df.to_numpy()
    global_means = Y_true.mean(axis=0, keepdims=True)  # shape (1, n_antibiotics)
    Y_pred = np.repeat(global_means, len(Y_true), axis=0)

    r2 = skmetrics.r2_score(Y_true, Y_pred, multioutput="uniform_average")
    rmse = np.sqrt(skmetrics.mean_squared_error(Y_true, Y_pred))
    return r2, rmse
```

```python
X, Y, feature_df, target_df = build_data_and_target(amr_df)

r2_persist, rmse_persist = baseline_persistence(feature_df, target_df)
print(f"Persistence baseline: R²={r2_persist:.3f}, RMSE={rmse_persist:.3f}")

r2_avg, rmse_avg = baseline_global_average(target_df)
print(f"Global average baseline: R²={r2_avg:.3f}, RMSE={rmse_avg:.3f}")
```

```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.multioutput import MultiOutputRegressor

def leave_one_pathogen_out_with_embeddings(
    amr_df,
    data_df,
    embedding_cols=None,
    n_estimators=200,
    random_state=42,
):
    """
    Leave-One-Pathogen-Out experiment with optional pathogen embeddings.

    - Train on all but one pathogen
    - Predict for the left-out pathogen
    - Rescale predictions back to raw counts
    - Return detailed DataFrame of results

    Args:
        amr_df (pd.DataFrame): must contain ["pathogen","antibiotic","start_date","count"]
        data_df (pd.DataFrame): must contain pathogen info + embedding columns.
        embedding_cols (list of str or None): Embedding columns to use.
                                              If None, only AMR features are used.
        n_estimators (int): number of trees in RandomForest
        random_state (int): random seed for reproducibility

    Returns:
        results_df (pd.DataFrame): detailed per-pathogen predictions
    """
    if embedding_cols:
        X, Y, feature_df, target_df, embeddings = build_amr_with_embeddings(
            amr_df, data_df, embedding_cols
        )
    else:
        X, Y, feature_df, target_df = build_data_and_target(amr_df)

    pathogens = feature_df.index
    antibiotics = [col.replace("_target", "") for col in target_df.columns]

    all_results = []

    for i, pathogen in enumerate(pathogens):
        # Train/test split
        X_train = np.delete(X, i, axis=0)
        Y_train = np.delete(Y, i, axis=0)
        X_test = X[i].reshape(1, -1)
        Y_test = Y[i].reshape(1, -1)

        # Train
        model = MultiOutputRegressor(
            RandomForestRegressor(
                n_estimators=n_estimators,
                random_state=random_state,
                n_jobs=-1
            )
        )
        model.fit(X_train, Y_train)
        Y_pred = model.predict(X_test)

        # Convert log1p back to raw counts
        actual_raw = np.expm1(Y_test.flatten())
        pred_raw = np.expm1(Y_pred.flatten())
        diff = pred_raw - actual_raw

        # Collect results in long format
        for ab, actual, pred, d in zip(antibiotics, actual_raw, pred_raw, diff):
            all_results.append({
                "pathogen": pathogen,
                "antibiotic": ab,
                "actual_count": actual,
                "predicted_count": pred,
                "difference": d
            })

        print(f"Done pathogen {i+1}/{len(pathogens)}: {pathogen}")

    results_df = pd.DataFrame(all_results)
    return results_df
```

```python
%%time
RUN_EXPERIMENT = False

if RUN_EXPERIMENT:
    results_df = leave_one_pathogen_out_with_embeddings(
        amr_df,
        data_df,
        embedding_cols=["taxonomy_embeddings", "anonymised_embedding"],
        n_estimators=300
    )
    results_df.to_parquet("results.parquet", index=False)
else:
    results_df = pd.read_parquet("results.parquet")

results_df
```
